## Probability

A **probability** is a numerical measure of the chance that an uncertain event will occur. Formally, an experiment with uncertain outcome is described by its **sample space** $\Omega$, the set of all possible outcomes, and an **event** $E$ is a subset of $\Omega$. The probability of $E$ is denoted $P(E)$.

Probabilities always lie in the closed interval $[0, 1]$:

$$
0 \leq P(E) \leq 1.
$$

The larger the probability, the more likely the event is to occur. A probability of $0$ indicates that the event will not occur, while a probability of $1$ indicates that it is certain to occur. Values strictly between $0$ and $1$ express intermediate degrees of chance.

When the sample space is finite and all outcomes are **equally likely**, the probability of an event is the proportion of outcomes that belong to it:

$$
P(E) = \frac{\text{number of outcomes favorable to } E}{\text{total number of possible outcomes}} = \frac{|E|}{|\Omega|},
$$

where $|\cdot|$ denotes the number of elements in a set. 

For example, when rolling a fair six-sided die, the event "the result is even" corresponds to $E = \{2, 4, 6\}$, so $P(E) = 3/6 = 1/2$.

> **Remark.** The statement that probability $0$ means "impossible" and probability $1$ means "certain" holds for finite sample spaces, but not in general. For a continuous random variable, such as a value drawn uniformly from $[0, 1]$, every individual outcome has probability $0$, yet one of them always occurs. In such cases, an event with probability $1$ is said to occur **almost surely**.

## General Probability Rules

The following rules hold for any events $A$, $B$, and $C$ defined on a sample space $\Omega$.

### Rule 1: Range of Probabilities

The probability of the impossible event $\emptyset$ is zero, and the probability of the certain event $\Omega$ is one. Every other event has a probability between these two extremes:

$$
0 \leq P(A) \leq 1.
$$

### Rule 2: Total Probability of the Sample Space

Since some outcome in $\Omega$ must occur, the sample space has probability one:

$$
P(\Omega) = 1.
$$

As a consequence, if the events $A_1, A_2, \ldots, A_m$ are mutually exclusive and together cover every possible outcome (that is, they form a **partition** of $\Omega$), their probabilities sum to one:

$$
\sum_{j=1}^{m} P(A_j) = 1.
$$

For example, if every voter must belong to exactly one of three political parties, Republican ($R$), Democrat ($D$), or Independent ($I$), then $P(R) + P(D) + P(I) = 1$.

### Rule 3: Complement Rule

The **complement** of $A$, denoted $A^{c}$, is the event that $A$ does not occur. Since $A$ and $A^{c}$ form a partition of $\Omega$, it follows from Rule 2 that

$$
P(A^{c}) = 1 - P(A), \qquad \text{equivalently} \qquad P(A) = 1 - P(A^{c}).
$$

### Rule 4: Addition Rule

The addition rule gives the probability that **at least one** of two events occurs, that is, $A$, $B$, or both.

(a) If $A$ and $B$ are **mutually exclusive** (disjoint), meaning they have no outcomes in common so that $A \cap B = \emptyset$, then

$$
P(A \cup B) = P(A) + P(B).
$$

(b) In general, when $A$ and $B$ may occur together, the outcomes in $A \cap B$ would be counted twice by the sum above, so their probability must be subtracted once:

$$
P(A \cup B) = P(A) + P(B) - P(A \cap B).
$$

Case (a) is a special case of (b), since $P(\emptyset) = 0$.

### Rule 5: Conditional Probability

The **conditional probability** of $A$ given $B$, denoted $P(A \mid B)$, is the probability that $A$ occurs given that $B$ is known to have occurred. Provided that $P(B) > 0$, it is defined as

$$
P(A \mid B) = \frac{P(A \cap B)}{P(B)}.
$$

Analogously, if $P(A) > 0$, then $P(B \mid A) = P(A \cap B) / P(A)$. Note that the vertical bar $\mid$ does not denote division. It is read as "given".

### Rule 6: Multiplication Rule

The multiplication rule gives the probability that **both** events occur. Rearranging the definition of conditional probability yields

$$
P(A \cap B) = P(A) \, P(B \mid A) = P(B) \, P(A \mid B),
$$

where both expressions are equal because $A \cap B = B \cap A$.

Two events are **independent** if the occurrence of one does not affect the probability of the other, that is, $P(A \mid B) = P(A)$. In this case the multiplication rule simplifies to

$$
P(A \cap B) = P(A) \, P(B).
$$

This extends to any number of **mutually independent** events. For instance, for three such events,

$$
P(A \cap B \cap C) = P(A) \, P(B) \, P(C).
$$

> **Remark.** Mutual independence of three or more events is a stronger requirement than pairwise independence. It is possible for each pair among $A$, $B$, and $C$ to be independent while the product formula above fails for all three together.

### Rule 7: The Chain Rule of Probability

The multiplication rule extends from two events to any number of events. Applying it repeatedly, the joint probability of events $A_1, A_2, \ldots, A_m$ can be decomposed into a product of conditional probabilities:

$$
P(A_1 \cap A_2 \cap \cdots \cap A_m) = P(A_1) \, P(A_2 \mid A_1) \, P(A_3 \mid A_1 \cap A_2) \cdots P(A_m \mid A_1 \cap \cdots \cap A_{m-1}),
$$

or, more compactly,

$$
P\left( \bigcap_{j=1}^{m} A_j \right) = \prod_{j=1}^{m} P\left( A_j \;\middle|\; \bigcap_{i=1}^{j-1} A_i \right),
$$

where the first factor is simply $P(A_1)$, since there is nothing to condition on. For three events, for example, this reads $P(A \cap B \cap C) = P(A) \, P(B \mid A) \, P(C \mid A \cap B)$. This identity, known as the **chain rule**, holds for any events with $P(A_1 \cap \cdots \cap A_{m-1}) > 0$ and requires no independence assumptions. The events may also be taken in any order, which yields different but equivalent factorizations. If the events are mutually independent, every conditional probability reduces to an unconditional one, and the chain rule simplifies to the product rule $P(A_1) \cdots P(A_m)$.

## Bayes' Theorem

Bayes' theorem expresses the conditional probability $P(A \mid B)$ in terms of the reverse conditional probability $P(B \mid A)$. It follows directly from the definition of conditional probability. Assuming that $P(A) > 0$ and $P(B) > 0$, we have

$$
P(A \mid B) = \frac{P(A \cap B)}{P(B)} \qquad \text{and} \qquad P(B \mid A) = \frac{P(A \cap B)}{P(A)}.
$$

Since $A \cap B = B \cap A$, both expressions can be solved for the same joint probability, which gives

$$
P(A \mid B) \, P(B) = P(A \cap B) = P(B \mid A) \, P(A).
$$

Dividing both sides by $P(B)$ yields **Bayes' theorem**:

$$
P(A \mid B) = \frac{P(B \mid A) \, P(A)}{P(B)}.
$$

Each term in this formula has a standard name. $P(A)$ is the **prior**, the probability of $A$ before observing $B$. $P(B \mid A)$ is the **likelihood** of observing $B$ when $A$ holds. $P(B)$ is the **evidence**, the overall probability of observing $B$. $P(A \mid B)$ is the **posterior**, the updated probability of $A$ after $B$ has been observed. Bayes' theorem thus describes how a prior belief is revised in light of new evidence.

In practice, $P(B)$ is often not known directly. It can be computed with the **law of total probability**, by splitting $\Omega$ into the partition $\{A, A^{c}\}$:

$$
P(B) = P(B \mid A) \, P(A) + P(B \mid A^{c}) \, P(A^{c}).
$$

## Naïve Bayes

### Probabilistic Model

Naïve Bayes is a **conditional probability model** for classification. A problem instance to be classified is represented by a feature vector $\mathbf{x} = (x_1, \ldots, x_d)$, and the model assigns to it a probability for each of $K$ possible classes $C_1, \ldots, C_K$:

$$
p(C_k \mid \mathbf{x}) = p(C_k \mid x_1, \ldots, x_d), \qquad k = 1, \ldots, K.
$$

Modeling this distribution directly quickly becomes infeasible. If the number of features $d$ is large, or if each feature can take many values, a table storing a probability for every combination of feature values would grow exponentially in size. The model is therefore reformulated to make it tractable. Applying Bayes' theorem, the conditional probability can be decomposed as

$$
p(C_k \mid \mathbf{x}) = \frac{p(C_k) \, p(\mathbf{x} \mid C_k)}{p(\mathbf{x})}.
$$

In the terminology of Bayesian probability, this reads

$$
\text{posterior} = \frac{\text{prior} \times \text{likelihood}}{\text{evidence}}.
$$

Only the numerator of this fraction matters for classification. The denominator $p(\mathbf{x})$ does not depend on the class $C_k$, and since the feature values are given, it is effectively a constant. The numerator equals the joint probability

$$
p(C_k) \, p(\mathbf{x} \mid C_k) = p(C_k, x_1, \ldots, x_d),
$$

which can be expanded by repeatedly applying the definition of conditional probability (the **chain rule**):

$$
\begin{aligned}
p(C_k, x_1, \ldots, x_d)
&= p(x_1 \mid x_2, \ldots, x_d, C_k) \, p(x_2, \ldots, x_d, C_k) \\
&= p(x_1 \mid x_2, \ldots, x_d, C_k) \, p(x_2 \mid x_3, \ldots, x_d, C_k) \, p(x_3, \ldots, x_d, C_k) \\
&= \cdots \\
&= p(x_1 \mid x_2, \ldots, x_d, C_k) \, p(x_2 \mid x_3, \ldots, x_d, C_k) \cdots p(x_d \mid C_k) \, p(C_k).
\end{aligned}
$$

### The Naïve Independence Assumption

At this point the "naïve" assumption is introduced: all features are assumed to be **mutually independent, conditional on the class** $C_k$. Under this assumption, each factor in the chain rule simplifies to

$$
p(x_j \mid x_{j+1}, \ldots, x_d, C_k) = p(x_j \mid C_k),
$$

so the joint model becomes

$$
p(C_k, x_1, \ldots, x_d) = p(C_k) \prod_{j=1}^{d} p(x_j \mid C_k).
$$

Consequently, the posterior distribution over the classes is

$$
p(C_k \mid x_1, \ldots, x_d) = \frac{1}{Z} \, p(C_k) \prod_{j=1}^{d} p(x_j \mid C_k) \;\propto\; p(C_k) \prod_{j=1}^{d} p(x_j \mid C_k),
$$

where $\propto$ denotes proportionality, and the evidence

$$
Z = p(\mathbf{x}) = \sum_{k'=1}^{K} p(C_{k'}) \, p(\mathbf{x} \mid C_{k'})
$$

is a scaling factor that depends only on $x_1, \ldots, x_d$. It is therefore constant once the feature values are known.

### The Naïve Bayes Classifier

The derivation above yields the independent feature model, that is, the **naïve Bayes probability model**. The **naïve Bayes classifier** combines this model with a decision rule. The most common rule is to select the most probable class, known as the **maximum a posteriori** (MAP) decision rule. The resulting classifier assigns the predicted label $\hat{y} = C_{\hat{k}}$, where

$$
\hat{k} = \underset{k \in \{1, \ldots, K\}}{\arg\max} \; p(C_k) \prod_{j=1}^{d} p(x_j \mid C_k).
$$

> **Remark.** In practice, the product of many probabilities can become too small to represent in floating-point arithmetic (numerical underflow). Since the logarithm is monotonically increasing, the classifier is usually computed in log space instead:
>
> $$
> \hat{k} = \underset{k}{\arg\max} \left[ \log p(C_k) + \sum_{j=1}^{d} \log p(x_j \mid C_k) \right].
> $$

## Zero Probability Problem I

When the probabilities in a Naïve Bayes classifier are estimated from data, they are computed as relative frequencies. For instance, $p(x_j = a \mid C_k)$ is estimated as the fraction of training examples of class $C_k$ in which feature $x_j$ takes the value $a$. This creates the **zero-probability problem**. If a particular feature value never occurs together with a given class in the training data, its estimated conditional probability is exactly $0$. Because the classifier multiplies these probabilities, a single zero factor makes the entire product zero, regardless of how strongly the remaining features support that class.

![bias](images/NaiveBayesZeroProbabilityProblem1.png)

In cases where such problems accur we use a technique called **Laplace Smoothing**

#### Laplace Smoothing

**Laplace smoothing** (also called additive smoothing) is a technique for smoothing categorical data that resolves this problem. It adds a small pseudo-count $\lambda > 0$ to every count, so that no probability estimate is ever zero. Let $n$ be the number of training examples, $N_k$ the number of examples belonging to class $C_k$, and $N_{k,j,a}$ the number of examples of class $C_k$ in which feature $x_j$ takes the value $a$. The smoothed estimates of the prior and conditional probabilities are then

$$
\hat{p}(C_k) = \frac{N_k + \lambda}{n + K\lambda},
\qquad
\hat{p}(x_j = a \mid C_k) = \frac{N_{k,j,a} + \lambda}{N_k + A_j \lambda},
$$

where $K$ is the number of distinct classes and $A_j$ is the number of distinct values that feature $x_j$ can take. The terms $K\lambda$ and $A_j\lambda$ in the denominators ensure that the smoothed probabilities still sum to one. The most common choice is $\lambda = 1$, which corresponds to **Laplace smoothing** in the strict sense. The general case $0 < \lambda < 1$ is known as **Lidstone smoothing**. Setting $\lambda = 0$ recovers the unsmoothed relative-frequency estimates.

## Zero Probability Problem II

## Numerical Underflow and Log Probabilities

The Naïve Bayes classifier computes a product of many probabilities, each of which lies between $0$ and $1$. As the number of features $d$ grows, this product shrinks rapidly toward zero. At some point the product will be rounded to exactly $0$, a phenomenon known as **numerical underflow**.

For example, with $d = 200$ features, each with conditional probability around $0.01$, the product is about $(10^{-2})^{200} = 10^{-400}$, which underflows to $0$ for every class. All classes then receive the same score, and the classifier can no longer distinguish between them.

![bias](images/NaiveBayesZeroProbabilityProblem2.png)


The standard solution is to work with **logarithms of probabilities**. Since $\log(ab) = \log a + \log b$, taking the logarithm turns the product into a sum of moderately sized negative numbers, which can be represented accurately:

$$
\log \left( p(C_k) \prod_{j=1}^{d} p(x_j \mid C_k) \right) = \log p(C_k) + \sum_{j=1}^{d} \log p(x_j \mid C_k).
$$

In the example above, the log score is about $200 \times \log(0.01) \approx -921$, which poses no numerical difficulty. Because the logarithm is a strictly increasing function, it preserves the ordering of the scores, so the class that maximizes the product also maximizes the sum. The MAP decision rule can therefore be applied directly in log space:

$$
\hat{k} = \underset{k}{\arg\max} \left[ \log p(C_k) + \sum_{j=1}^{d} \log p(x_j \mid C_k) \right].
$$

Note that the logarithm of zero is undefined ($\log 0 = -\infty$). This is a further reason to apply Laplace smoothing before taking logarithms.

## Gaussian Naïve Bayes

The methods above assume categorical features, whose conditional probabilities can be estimated by counting. When a feature $x_j$ is **continuous**, counting is no longer possible. **Gaussian Naïve Bayes** instead assumes that, within each class $C_k$, the feature follows a normal distribution:

$$
p(x_j \mid C_k) = \frac{1}{\sqrt{2\pi\sigma_{jk}^{2}}} \exp\left( -\frac{(x_j - \mu_{jk})^{2}}{2\sigma_{jk}^{2}} \right),
$$

where the mean $\mu_{jk}$ and variance $\sigma_{jk}^{2}$ are estimated from the training examples of class $C_k$, as the sample mean and variance of feature $x_j$. These densities are then substituted into the same decision rule as before, preferably in log space. Note that $p(x_j \mid C_k)$ is now a probability density rather than a probability, so it may exceed $1$. This does not affect classification, since only the relative scores of the classes matter.
