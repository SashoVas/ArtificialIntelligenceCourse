# Python and Notebooks

A Python notebook (usually a Jupyter Notebook) is an interactive document where you can write and run Python code in small sections called cells.

They are commonly used for:

- Data analysis - reading, cleaning, and exploring data.
- Visualization - creating charts and graphs.
- Machine learning/AI - training models and testing algorithms step by step.
- Experimentation - quickly trying code and seeing the results immediately.
- Documentation - combining code, output, explanations, formulas, and images in one document.
- Data engineering - testing transformations, querying data, and working with platforms such as Microsoft Fabric.

You can run notebooks locally or on Google servers - 
https://colab.research.google.com/
